```python
import pandas as pd 

BASE = '../../new-dataset/tennis_MatchChartingProject-master'

points_2010s = pd.read_csv(f'{BASE}/charting-m-points-2010s.csv')
points_2020s = pd.read_csv(f'{BASE}/charting-m-points-2020s.csv')
points_to_2009 = pd.read_csv(f'{BASE}/charting-m-points-to-2009.csv')

points = pd.concat([points_to_2009, points_2010s, points_2020s])
```

Para ver quem ganhou cada partida, é só ver quem marcou o último ponto.

```python
winners = points.sort_values(by=['match_id', 'Pt']).groupby('match_id').last().reset_index()[['match_id', 'PtWinner']].rename(columns={'PtWinner' : 'Winner'})
```

```python
winners.to_csv(f'{BASE}/winners.csv', index=False)
print('Salvo')
```
