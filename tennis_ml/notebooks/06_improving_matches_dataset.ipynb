```python
import pandas as pd
import numpy as np

BASE = '../../new-dataset/tennis_MatchChartingProject-master'

matches         = pd.read_csv(f'{BASE}/charting-m-matches.csv')
winners         = pd.read_csv(f'{BASE}/winners.csv')
player_profiles = pd.read_csv(f'{BASE}/player_profiles.csv')

matches['Date'] = pd.to_datetime(matches['Date'], format='%Y%m%d', errors='coerce')
matches = matches.merge(winners, how='left')

matches['Pl 1 hand'] = matches['Pl 1 hand'].where(matches['Pl 1 hand'].isin(['L', 'R']), other=np.nan)
matches['Pl 2 hand'] = matches['Pl 2 hand'].where(matches['Pl 2 hand'].isin(['L', 'R']), other=np.nan)

print('matches:         ', matches.shape)
print('player_profiles: ', player_profiles.shape)
```

```python
profile_cols = [c for c in player_profiles.columns if c not in ['match_id', 'player', 'Date', 'Winner']]

p1_profiles = (player_profiles[['match_id', 'player'] + profile_cols]
    .rename(columns={c: f'p1_{c}' for c in profile_cols})
    .rename(columns={'player': 'Player 1'})
)

p2_profiles = (player_profiles[['match_id', 'player'] + profile_cols]
    .rename(columns={c: f'p2_{c}' for c in profile_cols})
    .rename(columns={'player': 'Player 2'})
)

dataset = (matches
    .merge(p1_profiles, on=['match_id', 'Player 1'], how='left')
    .merge(p2_profiles, on=['match_id', 'Player 2'], how='left')
)

print(dataset.shape)
dataset.head()
```

```python
dataset.to_csv(f'{BASE}/matches_dataset.csv', index=False)
```
